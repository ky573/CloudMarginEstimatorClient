# Margin Estimator Tool Tutorial: Using the `get_products` Feature

This tutorial demonstrates how to use the `get_products` command from the Margin Estimator Tool. This command allows you to retrieve product information from the API with various filtering and export options.

## Introduction

The `get_products` command fetches product data from the products endpoint. It supports:
- Specifying the business date
- Choosing between SOD (Start of Day) and LIVE versions
- Specifying the timestamp for LIVE versions
- Exporting results to Excel, JSON, or default CSV
- Applying various filters to get specific products

Let's explore how to use this feature with some examples.

## Setup

First, let's assume we have the Margin Estimator Tool installed. We'll use the command line interface directly from Jupyter using the `!` prefix.

```python
# Display help to see available commands
!margin_estimator_tool --help
```

```python
# Display help for the get_products command
!margin_estimator_tool get_products --help
```

## Example 1: Basic Usage - Get All Products for Today

Let's start with a simple example to fetch all products for the current date and display them in the console.
For the today's date to be considered, we need to specify the version to be LIVE.

```python
# Get all products for today (default)
!margin_estimator_tool get_products --version LIVE
```

By default, the command fetches all products for the current date. The output first shows some information from logger regarding the request.
Then it displays that the request was successful and directory where products were exported (default to CSV file).

## Example 2: Export Products to Excel using historic date

Now, let's fetch products for a specific date and export them to an Excel file.
Specifying a historic date will make us wait for the response for up to 5 minutes, so we need to keep that in mind.


```python
# Get products for a specific date and export to Excel
!margin_estimator_tool get_products --date 20250228 --to_excel
```

In this example, we:

1. Specified a date (February 28, 2025) using `--date 20250228`
2. Requested an Excel export with `--to_excel`

The output shows information from the logger, then that the request was successful, and created an Excel file in the specified directory.

## Example 3: Using Filters

Let's filter products by currency and product type.

```python
# Get EUR currency option index products
!margin_estimator_tool get_products --filter currency:EUR,product_type:OINX
```

In this example, we used the `--filter` parameter to filter products that have:
- Currency: EUR
- Product type: OINX (Option Index)

The output shows that 80 products matched these criteria.

## Example 4: Using LIVE Version with Timestamp

Now let's get the LIVE version of products with a specific timestamp and export to JSON.

```python
# Get LIVE version products with timestamp and export to JSON
!margin_estimator_tool get_products --version LIVE --timestamp 1709460000000 --to_json
```

In this example, we:

1. Specified the LIVE version with `--version LIVE`
2. Provided a specific timestamp with `--timestamp 1709460000000`
3. Requested a JSON export with `--to_json`

The output shows the request was successful, and created a JSON file in project's root directory.

## Example 5: Multiple Filters with Excel Export and specified directory using absolute path

Let's combine multiple filters and export options.

```python
# Get products with multiple filters and export to Excel
!margin_estimator_tool get_products \
  --filter clearing_house:EUXCDEFF,currency:CHF,product_settlement_type:CASH \
  --to_excel \
  --export_dir /home/gk101/workspace
```

In this example, we combined multiple filters to get products that have:
- Clearing house: EUXCDEFF
- Currency: CHF
- Settlement type: CASH

The output shows that 110 products matched these criteria, and they were exported to an Excel file to desired directory.

## Summary

The `get_products` command in the Margin Estimator Tool provides a flexible way to retrieve and filter product data. Key features include:

- **Date selection**: Specify a business date or use defaults (SOD = previous business day, LIVE = current day)
- **Version options**: Choose between SOD (Start of Day) and LIVE versions
- **Export formats**: Export CSV, Excel, or JSON files
- **Filtering**: Apply filters to narrow down results based on product characteristics
- **Timestamping**: For LIVE versions, specify a timestamp to get data at a particular point in time

This command is particularly useful for:
- Getting product reference data for margin calculations
- Analyzing available products in the system
- Preparing data for further processing in ETD portfolio analysis
- Monitoring product changes between different snapshots

Remember to check the documentation or use the `--help` option for detailed information about all available parameters.
