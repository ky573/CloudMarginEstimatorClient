# Margin Estimator Tool Tutorial: Using the `get_snapshots` Command

This tutorial demonstrates how to use the `get_snapshots` command of the margin estimator tool to retrieve information about available EOD (End of Day) snapshots for a specified date range.

## Introduction

Snapshots in the margin estimator tool represent point-in-time captures of market data. The `get_snapshots` command specifically retrieves EOD (non-live) snapshots, which reflect the state of the market at the end of each trading day.

The `get_snapshots` command requires at least one parameter: `--date_from` to specify the starting date of the range to query. Optionally, you can provide `--date_to` to specify the end date (defaults to the current date if not provided).

Let's see the options proided to us:

```python
# Display help
!margin_estimator_tool get_snapshots --help
```

## Basic Usage with Default `date_to`

When not specifying `date_to` parameter, it defaults to current date, so that means that the latest EOD snapshot we will get will be from yesterday.

```python
!margin_estimator_tool get_snapshots --date_from 20250303
```

### Understanding the Output

The output shows a list of available EOD snapshots for the specified date range. For each snapshot, the following information is displayed:

- date: The business date of the snapshot in YYYYMMDD format
- OTC: Whether OTC (Over-The-Counter) data is available (YES/NO)
- CASH: Whether cash data is available (YES/NO)

## Get Snapshots for a Single Date

You can also retrieve snapshots for a single date by setting both `--date_from` and `--date_to` to the same value:

```python
!margin_estimator_tool get_snapshots --date_from 20250303 --date_to 20250303
```

## Specifying Whole Range

Let's now specify both the start date and the end date. We will get results for all business days in that range.

```python
!margin_estimator_tool get_snapshots --date_from 20250202 --date_to 20250303
```

## Use Cases

The `get_snapshots` command is particularly useful for:

1. Data Availability Checking: Before running analyses that depend on snapshot data, we can check which dates have available snapshots

2. Verifying Data Completeness: We can check if both OTC and CASH data are available for a particular date

3. Historical Analysis: When performing historical analyses, we can first determine which historical dates have snapshot data available

## Error Handling

The command implements error handling for various scenarios. Let's look at some common error cases:

```python
# Invalid date format
!margin_estimator_tool get_snapshots --date_from 2025-03-03
```

```python
# Earlier end date than start date
!margin_estimator_tool get_snapshots --date_from 20250306 --date_to 20250303
```

Here, the result will be empty request sent back to the user.

## Conclusion

The `get_snapshots` command is a powerful tool for retrieving information about available SOD snapshots. Key points to remember:

- Use `--date_from` to specify the start date (required)
- Use `--date_to` to specify the end date (optional, defaults to current date)
- The command returns only non-live (SOD) snapshots
- For each snapshot, it shows the date and availability of OTC and CASH data

This command is particularly useful for checking data availability before performing margin calculations or other analyses that depend on snapshot data.
