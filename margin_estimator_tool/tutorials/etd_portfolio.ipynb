# Margin Estimator Tool Tutorial: Using the `etd_portfolio` Feature

This tutorial demonstrates how to use the `etd_portfolio` command from the margin estimator tool. The command allows you to upload an ETD (Exchange-Traded Derivatives) portfolio from a CSV file and process it through the estimator API.

## Introduction

The `etd_portfolio` command is part of the margin estimator CLI tool. It validates and processes ETD portfolio data, then outputs the results in your desired format (CSV, Excel, or JSON).

The `etd_portfolio` command accepts the following parameters:

- `--csv_file` (Required): Path to the ETD portfolio CSV file
- `--date`: Specific business date in YYYYMMDD format
- `--version`: Snapshot version ("SOD" or "LIVE")
- `--timestamp`: Timestamp for LIVE version
- `--to_excel`: Flag to export results to an Excel file
- `--to_json`: Flag to export results to a JSON file
- `--export_dir`: Output directory for exported files

## CSV File Requirements

Your CSV file must have one of two supported header formats:

1. GUI Format:
```
Product ID,Contract Date,Call Put Flag,Exercise Price,Version Number,Net LS Balance
```

2. Inner Format:
```
call_put_flag,component_margin,component_margin_currency,contract_date,exercise_price,exercise_style,iid,instrument_type,line_no,liquidation_group,liquidation_group_split,maturity,net_ls_balance,premium_margin,premium_margin_currency,product_id,version_number
```

The tool validates these headers before processing.

## Using Template from `get_series` feature for ETD Portfolio

Let's now try to generate our template for ETD portfolio from series endpoint. The only think we will need to do is to manually define the positions, otherwise they will dafault to 0.

````python
!margin_estimator_tool get_series \
    --version LIVE \
    --products BMW \
    --export_dir ../data \
    --template
````

Let's see how the file looks like:

````python
with open("../data/20250310_LIVE_etd_portfolio_template.csv", "r") as f:
    print(f.read())
````

Now, let's enter our positions (here we do it using little scripting):

````python
import csv

# Read the CSV file and modify the rows
with open("../data/20250310_LIVE_etd_portfolio_template.csv", newline="", mode="r") as f:
    reader = csv.reader(f)
    rows = list(reader)  # Convert reader to a list
    header = rows[0]  # Extract header
    data_rows = [row + ["100"] for row in rows[1:]]  # Append 100 only to data rows

# Write the modified data back to the same file
with open("../data/20250310_LIVE_etd_portfolio_template.csv", newline="", mode="w") as f:
    writer = csv.writer(f)
    writer.writerow(header)  # Write header without modification
    writer.writerows(data_rows)  # Write modified data rows
````

Now, our portfolio looks like this:

````python
with open("../data/20250310_LIVE_etd_portfolio_template.csv", "r") as f:
    print(f.read())
````

Let's use it in our `etd_portfolio` feature and export to JSON:

````python
!margin_estimator_tool etd_portfolio \
    --version LIVE \
    --csv_file ../data/estimator_etd_csv.csv \
    --to_json
````

## Error Handling

Here are some common issues you might encounter when using the `etd_portfolio` command and how to resolve them:

1. **Invalid Headers**:
   - Error: `Headers mismatch. Expected: either 'Product ID,...' or 'call_put_flag,...'`
   - Solution: Ensure your CSV file has exactly one of the two supported header formats


````python
# Error in the GUI header, missing last field 'Net LS Balance'
!margin_estimator_tool etd_portfolio \
    --csv_file ../data/estimator_etd_csv.csv \
    --version LIVE
````

Similar message will get printed out in case of a wrong header for Inner format.

2. **Missing Required Fields**:
   - Error: Missing required field (like Product ID in GUI header)
   - Solution: Make sure all required fields have values in each row


````python
# Required field has no value
!margin_estimator_tool etd_portfolio \
    --csv_file ../data/estimator_etd_csv.csv \
    --version LIVE
````
