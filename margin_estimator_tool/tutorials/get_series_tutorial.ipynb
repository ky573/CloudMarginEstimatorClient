# Margin Estimator Tool Tutorial: Using the `get_series` Feature

This tutorial demonstrates how to use the `get_series` command from the margin estimator tool to fetch and filter series data. We'll explore various parameters and options to help us effectively use this feature.

## Introduction

The `get_series` command allows us to fetch financial instrument series data for specific products from the series endpoint. It supports:

- Filtering by type (option/future), call/put flag
- Filtering by time to expiration (TTE)
- Specifying product(s) (mandatory)
- Specifying business date (defaults to current date in YYYYMMDD form), version (LIVE or EOD), and timestamp for LIVE version
- Generating ETD portfolio templates
- Export to Excel, JSON, or  (default) CSV formats. Excel and JSON are mutually exclusive
- Apply custom filters using key-value pairs

Let's now check options vailable to us:

```python
# Display help
!margin_estimator_tool get_series --help
```

## Basic Usage for the Current Day

Let's start with a basic example of fetching series data for a specific product. We need to use the version LIVE, if the request for today's date should be successful.

```python
# Basic usage - fetch series data for BMW product
!margin_estimator_tool get_series \
    --products BMW \
    --version LIVE
```

This command fetches all series data for the BMW product, and exports it as a CSV file to the default directory. By default, without specifying `--to_excel` or `--to_json`, the data is exported as CSV.


## Sample Output

Here's what the exported CSV might look like:

```python
# Display the output file (this is just for demonstration)
import pandas as pd
pd.read_csv("/path to file/").head(3)
```

## Fetching multiple Products with Historic Date

You can fetch series data for multiple products by separating them with commas.
Let's also use the historic date in YYYYMMDD format, but we need to keep in mind that this kind of request may take up to 5 minutes to complete.

```python
# Fetch series for multiple products and historic request
!margin_estimator_tool get_series \
    --products BMW,DAI \
    --date 20250228
```

## Filtering by Series Type (Option/Future)

You can filter series data by type using the `--type` parameter:

```python
# Filter by type (option) and call_put_flag (C)
!margin_estimator_tool get_series \
    --products BMW \
    --type option \
    --version LIVE
```

```python
# Filter by type (future)
!margin_estimator_tool get_series \
    --products BMW \
    --type future \
    --version LIVE
```

## Filtering by Call/Put Flag

For option series, you can filter by call or put flag:

```python
# Filter by call flag
!margin_estimator_tool get_series \
    --products BMW \
    --call_put_flag C \
    --version LIVE
```

```python
# Filter by put flag
!margin_estimator_tool get_series \
    --products BMW \
    --call_put_flag P \
    --version LIVE
```

## Filtering by Time to Expiration (TTE)

You can filter series by their days to expiration using `--min_tte` and `--max_tte` parameters. Here, we will define the interval, but we can also choose only one of the filters:

```python
# Filter by time to expiration (short-term options only)
!margin_estimator_tool get_series \
    --products BMW \
    --max_tte 30 \
    --min_tte 10 \
    --version LIVE
```

## Using Custom Filters

You can apply more specific filters using the `--filter` parameter with key-value pairs:

```python
# Filter by specific exercise price
!margin_estimator_tool get_series \
    --products BMW \
    --filter "exercise_price:91" \
    --version LIVE
```

```python
# Multiple filters combined with comma
!margin_estimator_tool get_series \
    --products BMW \
    --filter "exercise_price:91,call_put_flag:P,contract_date:20240315" \
    --version LIVE
```

In this case, there were no series found with such filters.

## Exporting to Different Formats

You can export the data to Excel or JSON formats:

```python
# Export to Excel
!margin_estimator_tool get_series \
    --products BMW \
    --to_excel \
    --version LIVE
```

```python
# Export to JSON
!margin_estimator_tool get_series \
    --products BMW \
    --to_json \
    --version LIVE
```

## Generating an ETD Portfolio Template

You can generate a template for an ETD portfolio based on the filtered series:

```python
# Generate a template for an ETD portfolio
!margin_estimator_tool get_series \
    --products BMW \
    --filter "exercise_price:91" \
    --template \
    --version LIVE
```

The generated template will look like this:

```python
# Display the template (for demonstration)
pd.read_csv("/path to template/").head(3)
```

The user will manually need to enter positions into Net LS Balance, otherwise it will default to 0.

## Error Handling

The tool provides error messages for various scenarios:

```python
# Error when using an invalid filter field
!margin_estimator_tool get_series \
    --products BMW \
    --filter "unknown_field:value" \
    --version LIVE
```

```python
# Error for invalid numeric value
!margin_estimator_tool get_series \
    --products BMW \
    --filter "exercise_price:string_value" \
    --version LIVE
```

## Understanding the Series Data

Here's an explanation of the key fields in the series data:

| Field | Description |
|-------|-------------|
| `product_id` | Identifier for the financial product (e.g., BMW) |
| `contract_date` | Date of the contract expiration (YYYYMMDD) |
| `call_put_flag` | 'C' for call options, 'P' for put options, null for futures |
| `exercise_price` | Strike price for options |
| `version_number` | Version of the contract |
| `iid` | Unique identifier for the instrument |
| `act_trade_unit_no` | Number of units in a standard trade |
| `days_to_expiration` | Days remaining until contract expiration |
| `contract_maturity` | Contract maturity month and year (YYYYMM) |
| `expiry_maturity` | Expiration maturity month and year (YYYYMM) |
| `trade_unit_value` | Value per trade unit |
| `exercise_style_flag` | 'A' for American, 'E' for European style options |
| `contract_frequency` | Frequency of the contract (e.g., MONTHLY) |

## Summary

The `get_series` command is a tool for fetching and filtering financial instrument series data. By combining various parameters, we can precisely select the data we need and export it in our preferred format.

This data can be used for further analysis or as input to other components of the margin estimator tool, such as the margin calculator.

Remember to check the documentation or use the `--help` option for detailed information about all available parameters.
