## Introduction to Calendar spread strategy

A calendar spread is an options strategy that involves buying and selling options with the same strike price but different expiration dates. Specifically, a long calendar spread is established by buying a longer-term option and selling a shorter-term one. This strategy benefits from the difference in time decay (theta) between the two options, with the short-term option's value eroding faster, especially if the underlying asset's price stays near the strike price.

One of the main advantages of a long calendar spread is its defined risk profile: the maximum loss is limited to the initial net debit (the upfront cost of opening the position). This makes the strategy more capital efficient compared to purchasing only a long-term option.

From a margin perspective, calendar spreads offer favorable treatment. The long option acts as a hedge against the short option, reducing the broker's required margin compared to holding a naked short position. The margin requirement is usually tied to the initial cost of the spread.

Additionally, there is assignment risk for the short-term option as it nears expiration, especially if it moves in-the-money. However, the long-term option provides a layer of protection, as it can offset the obligations resulting from the assignment.

In summary, long calendar spreads are a risk-defined, capital-efficient strategy that offers reduced margin requirements compared to uncovered short positions, while still allowing for profits if the underlying asset's price remains near the strike as the short option expires.

## Output Interpretation

Let's now analyze the output. We will first load the portfolio.

```python
import json
import pandas as pd

with open('../output/20250409_LIVE_estimator.json', 'r') as file:
    data = json.load(file)

# Extract portfolio margin details
portfolio_margin = data["portfolio_margin"][0]

initial_margin = portfolio_margin['initial_margin']
market_risk = portfolio_margin['market_risk']
premium_margin = portfolio_margin['premium_margin']

# Display the portfolio margin summary
portfolio_margin_summary = {
    "Initial Margin": initial_margin,
    "Market Risk": market_risk,
    "Premium Margin": premium_margin,
}

portfolio_margin_summary_df = pd.DataFrame(portfolio_margin_summary, index=[0])
portfolio_margin_summary_df
```

```python
# Extract and display the drilldown data for each instrument
drilldown_data = data["drilldowns"]
drilldown_list = []

for d in drilldown_data:
    drilldown_list.append({
        "Product ID": d["product_id"],
        "Contract Date": d["contract_date"],
        "Type": "Call" if d["call_put_flag"] == "C" else "Put",
        "Exercise Price": d["exercise_price"],
        "Position": "Long" if d["net_ls_balance"] > 0 else "Short",
        "Component Margin": d["component_margin"],
        "Premium Margin": d["premium_margin"]
    })

drilldown_df = pd.DataFrame(drilldown_list)
drilldown_df
```

## Conclusion

In a calendar spread, the margin analysis highlights the effects of the short-term sold option and the long-term bought option. The negative margin for the long-term option is offsetting the margin requirement for the short-term option, reducing the overall margin for the entire calendar spread. 

This structure of the portfolio helps manage risk and capital efficiency, with the premium margin indicating the income from selling the short-term option. The net margin reflects the hedged nature of the position, with an overall reduced margin requirement compared to outright options trades.
