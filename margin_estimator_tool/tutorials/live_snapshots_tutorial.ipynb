# Margin Estimator Tool Tutorial: Using the `get_live_snapshots` Feature

This tutorial demonstrates how to use the `get_live_snapshots` command from the margin estimator tool. This command allows us to retrieve information about available live snapshots for a specific business date.

## Introduction

Live snapshots represent point-in-time records of market data that can be used for margin calculations. Each snapshot has properties such as:
- Timestamp (when the snapshot was taken)
- OTC availability status
- Cash availability status

These snapshots can be used for further analysis or as inputs to other commands like `margin_calculator` or `etd_portfolio`.

Let's see options available to us:

```python
# Display help
!margin_estimator_tool get_live_snapshots --help
```

## Basic Usage for the Current Date

Let's start with the basic example, getting the data for the current date. The date always needs to be specified.

```python
!margin_estimator_tool get_live_snapshots --date 20250307
```

## Understanding the Output

The command displays:
- A header showing the business date
- A numbered list of available snapshots
- For each snapshot:
  - Time in HH:MM:SS format
  - Timestamp in milliseconds (Unix epoch)
  - OTC (Over-The-Counter) availability (YES/NO)
  - CASH availability (YES/NO)

This information can be used to:
1. Determine suitable points in time for analysis
2. Choose specific timestamps for other commands
3. Verify data availability (OTC and CASH)

## Historic Date

Let's fetch live snapshots for some historic date. They might take longer to return.

When there are no available snapshots (i.e., not a business day), we will get notified from the request.
